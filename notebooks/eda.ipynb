```python
import numpy as np

df['Mechanical Power [W]'] = (
    df['Torque [Nm]']
    * df['Rotational speed [rpm]']
    * 2 * np.pi / 60
)
```

Machine failure class the target variable is highly imbalenced that is eventually while training the model it would become lazy and say all the machines are good forgetting to detect the failure ones i.e it does not cares about the true negative values but the accuracy will be 96 which sounds good but dosent not help in neither way so accuracy only is not enough metric in the model 

Metrics such as precision, recall, F1-score, and ROC-AUC/PR-AUC should be considered when evaluating the model.
